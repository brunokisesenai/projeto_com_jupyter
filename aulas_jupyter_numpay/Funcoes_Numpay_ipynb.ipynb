```python
vendas = np.array([
    [15000, 20000, 30000], #1º trimestre
    [25000, 30000, 40000], #2º trimestre
    [35000, 40000, 50000], #3º trimestre
    [45000, 50000, 60000], #4º trimestre
])

#total_vendas_ano = sum(vendas)
total_vendas_ano = np.sum(vendas)
print(f"Total de vendas: {total_vendas_ano}")

total_por_trimestre = np.sum(vendas, axis=1)
print(f"Total de por trimestre: {total_por_trimestre}")

comeco_de_trimestre = np.sum(vendas, axis=0)
print(f"Total do começo do trimestre: {comeco_de_trimestre}")

maior_venda_ano = np.max(vendas)
print(f"Maior venda do ano: {maior_venda_ano}")
menor_venda_ano = np.min(vendas)
print(f"Menor venda do ano: {menor_venda_ano}")
```

```python
valor_produtos = np.array([
    10,15,20,30
])

#np.where (condição, True, False)
valor_promocao = np.where(valor_produtos >= 20, valor_produtos * 0.9, valor_produtos)

valor_promocao = np.round(valor_promocao, 2)
print(valor_promocao)
```

#Lidando com o Numpy
O Numpy posibilita cria arrays, que são listas onde podemos fazer lógicas e operações booleanas dentro de cada valor da lista.
Isso significa que o python passa a tratar os valores como unitários.

```python
import pandas as pd

df_produtos = pd.DataFrame(
    {
        "Produto": ['Notebook', 'Celular', 'Monitor'],
        "Estoque": [10, 0, 5],
        "Valor": [2000, 1500, 500],
        "Cliente": ["Magalu", "Americanas", "Kabum"]
    }
)


df_produtos["Valor"] = np.where(
    (df_produtos["Valor"] > 1500),
    df_produtos["Valor"] * 0.9,
    df_produtos["Valor"]
)

df_produtos["Valor"] = np.where(
    df_produtos["Cliente"] == "Kabum",
    df_produtos["Valor"] * 0.8,
    df_produtos["Valor"]
)

df_produtos["Estoque"] = np.where(
    df_produtos["Cliente"] == "Kabum",
    df_produtos["Estoque"] == 0,
    np.where(
        df_produtos["Cliente"] == "Magalu",
        df_produtos["Estoque"] - 2,
        df_produtos["Estoque"]
    )
)

df_produtos["Status"] = np.where(
    (df_produtos["Estoque"] > 0),
    "Disponível",
    "Esgotado"
)


print(df_produtos)
```

```python
agrupamento = df_produtos.groupby(["Produto", "Estoque"])['Valor'].sum()

print(agrupamento)

agrupamento_count = df_produtos.groupby(["Cliente", "Estoque"])["Valor"].sum()

print(agrupamento_count)
```
