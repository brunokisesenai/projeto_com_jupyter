```python
import numpy as np
import pandas as pd

mercados_df = pd.read_csv('vendas_mercados.csv')

display(mercados_df)




```

RESOLVENDO A PRIMEIRA QUESTÃO


```python
mercados_df["lucro"] = ((
    mercados_df["Preco_Unitario"] - mercados_df["Custo_Unitario"])
    * mercados_df["Quantidade"]
)

lucro_por_mercado = mercados_df.groupby(["Mercado"])["lucro"].sum()
maior_lucro = lucro_por_mercado.max()

print(lucro_por_mercado)
print(f"O maior lucro: {maior_lucro}")
```

RESOLVENDO A SEGUNDA QUESTÃO

```python
pagamento_por_regiao = mercados_df.groupby(["Regiao", "Forma_Pagamento"])["ID_Venda"].count()


print(pagamento_por_regiao)

```

RESOLVENDO A TERCEIRA QUESTÃO

```python
vendas_por_produto = mercados_df.groupby('Produto')['Quantidade'].sum()
print(vendas_por_produto)

produto_mais_vendido = vendas_por_produto.idxmax()
valor_mais_vendido = vendas_por_produto.max()
produto_menos_vendido = vendas_por_produto.idxmin()
valor_menos_vendido = vendas_por_produto.min()

print(f"O produto mais vendido foi: {produto_mais_vendido} com {valor_mais_vendido} unidades")
print(f"O produto_menos_vendido foi: {produto_menos_vendido} com {valor_menos_vendido} unidades")

display(vendas_por_produto.sort_values(ascending=False))
```

RESOLVENDO A QUARTA QUESTÃO

```python
media_mercados = mercados_df.groupby('Mercado')['Preco_Unitario'].mean()

media_mercados = media_mercados.sort_values(ascending=False)


print("PREÇO MÉDIO DOS PRODUTOS POR MERCADO:")
print(media_mercados.round(2))
```

RESOLVENDO A QUINTA QUESTÃO

```python
mercados_df['Faturamento'] = mercados_df['Preco_Unitario'] * mercados_df['Quantidade']


faturamento_total = mercados_df['Faturamento'].sum()


faturamento_mercados = mercados_df.groupby('Mercado')['Faturamento'].sum()

faturamento_mercados = faturamento_mercados.sort_values(
    ascending=False
)

maior_receita = faturamento_mercados.idxmax()
valor_maior_receita = faturamento_mercados.max()


print("FATURAMENTO TOTAL DA REDE:")
print(f"R$ {faturamento_total:,.2f}")

print("\nFATURAMENTO POR MERCADO:")
print(faturamento_mercados)

print("\nMERCADO COM MAIOR RECEITA:")
print(maior_receita)
print(f"Faturamento: R$ {valor_maior_receita:,.2f}")


```

RESOLVENDO A SEXTA QUESTÃO

```python

mercados_df['Preco_Com_Desconto'] = np.where(
    mercados_df['Quantidade'] >= 10,
    mercados_df['Preco_Unitario'] * 0.90,
    mercados_df['Preco_Unitario']
)


mercados_df['Novo_Faturamento'] = (
    mercados_df['Preco_Com_Desconto'] * mercados_df['Quantidade']
)


novo_faturamento_total = mercados_df['Novo_Faturamento'].sum()



faturamento_original = (
    mercados_df['Preco_Unitario'] * mercados_df['Quantidade']
).sum()


novo_faturamento = mercados_df['Novo_Faturamento'].sum()


valor_desconto = faturamento_original - novo_faturamento

print(f"Faturamento original: R$ {faturamento_original:,.2f}")
print(f"Novo faturamento: R$ {novo_faturamento:,.2f}")
print(f"Valor concedido em descontos: R$ {valor_desconto:,.2f}")
```
