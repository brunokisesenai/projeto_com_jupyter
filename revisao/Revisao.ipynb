TYPE HINTING

```python
#SEM O TYPE HINTING:
def calculo_media(notas):
    return sum(notas) / len(notas)

#COM O TYPE HINTING::
def calculo_media_tipada(notas: list[float]) -> float:
    return sum(notas) / len(notas)


notas = [10.0, 8.5, 9.6, 6.8]

calculo_media(notas)
print(calculo_media(notas))
print(calculo_media_tipada(notas))
```

DEBUGGING (DEPURAÇÃO)

Geralmente utilizada para identificar problemas

a) Breakpoint: ponto de parada

b) Step Over: passar por cima (ele pula uma linha ou vai para o próximo breakpoint).

c) Step In: entra na função.


d) Step Out: Sai da função

```python
contador = 0

while contador < 10:
    print(contador)
    contador += 1


for index, nota in enumerate(notas):
    print(f"{index}: {nota:.2f}")
```
