TYPE HINTING

```python
#SEM O TYPE HINTING:
def calculo_media(notas):
    return sum(notas) / len(notas)

#COM O TYPE HINTING::
def calculo_media_tipada(notas: list[float]) -> float:
    return sum(notas) / len(notas)


notas = [10.0, 8.5, 9.6, 6.8]

calculo_media(notas)
print(calculo_media(notas))
print(calculo_media_tipada(notas))
```

DEBUGGING (DEPURAÇÃO)

Geralmente utilizada para identificar problemas

a) Breakpoint: ponto de parada

b) Step Over: passar por cima (ele pula uma linha ou vai para o próximo breakpoint).

c) Step In: entra na função.


d) Step Out: Sai da função

```python
contador = 0

while contador < 10:
    print(contador)
    contador += 1


for index, nota in enumerate(notas):
    print(f"{index}: {nota:.2f}")
```

LIST COMPREHENSION (COMPRESSÃO DE LISTA)

```python
#SEM COMPRESSÃO

temperaturas = [-10, 18, 30, -90, 50]
temperaturas_validas = []

for temperatura in temperaturas:
    if temperatura > 0:
        fahrenheit = (temperatura * 1.0) + 32

        temperaturas_validas.append(fahrenheit)

print(temperaturas_validas)
```

```python
#COM COMPRESSÃO

temperaturas = [-10, 18, 30, -90, 50]
temperaturas_validas = [
    (temperatura * 1.0) + 32
    for temperatura in temperaturas
    if temperatura > 0
]

print(temperaturas_validas)
```

DECORATOR (Decoradores)

```python
def meu_decorador(funcao):
    def executar():
        print("="*20)
        print("\nInciando o programa...\n")
        funcao()
        print("\nFinalizando o programa...\n")
        print("="*20+"\n")
    return executar

@meu_decorador
def olaMundo():
    print("Olá Mundo!")

olaMundo()
```
